```python


# Step 1: Find the latest quarter
latest_quarter = State_txnsplit["Quarter"].max()

# Step 2: Filter for latest quarter
latest_Distribution = (
    State_txnsplit[State_txnsplit["Quarter"] == latest_quarter]
    .groupby(["State","Transaction Type"])
    .size()
    .reset_index(name="Transaction Counts")
)

# Step 3: Find top 5 states by total transaction counts
top_states = (
    latest_Distribution.groupby("State")["Transaction Counts"]
    .sum()
    .nlargest(5)
    .index
)

# Step 4: Filter data for only top 5 states
top5_Distribution = latest_Distribution[latest_Distribution["State"].isin(top_states)]

# Step 5: Plot bar chart
plt.figure(figsize=(10,6))
sns.barplot(
    data=top5_Distribution,
    x="State",
    y="Transaction Counts",
    hue="Transaction Type"
)

plt.xticks(rotation=45)
plt.title(f"Transaction Type Distribution - Top 5 States - Quarter {latest_quarter}")
plt.ylabel("Transaction Counts")
plt.xlabel("State")
plt.tight_layout()
plt.show()
```

DATA QUALITY CHECKS

```python
#calculating total umber of transaction ,total transaction amount and total registered users by summing up the values from district level data

Data_quality_district = District_txn_and_users.groupby("State").agg(total_no_Transaction=("Transactions","sum"),
                                                           total_transaction_amount=("Amount (INR)","sum"),
                                                           total_registered_users=("Registered Users","sum")).reset_index()
                                                           
Data_quality_district
Data_quality_state = State_transaction_and_users.groupby("State").agg(total_no_Transaction=("Transactions","sum"),
                                                           total_transaction_amount=("Amount (INR)","sum"),
                                                           total_registered_users=("Registered Users","sum")).reset_index()
                                                           
Data_quality_state
comparison = Data_quality_state.merge(Data_quality_district,on="State",how="outer",suffixes=("_state","_district"))
comparison
```

```python
comparison["transaction_diff"] = (
    comparison["total_no_Transaction_district"] -
    comparison["total_no_Transaction_state"]
)

comparison["amount_diff"] = (
    comparison["total_transaction_amount_district"] -
    comparison["total_transaction_amount_state"]
)

comparison["users_diff"] = (
    comparison["total_registered_users_district"] -
    comparison["total_registered_users_state"]
)
comparison["amount_diff"]=comparison["amount_diff"].astype("int64")

comparision
dq_issues = comparison[
    (comparison["transaction_diff"] != 0) |
    (comparison["amount_diff"] != 0) |
    (comparison["users_diff"] != 0)
]
#dq_issues["amount_diff"]=dq_issues["amount_diff"].astype("int64")
dq_issues
```

SUMMARY

Key Findings & Insight

1.Transaction & User Growth

*High transaction volumes are concentrated in large, urbanized states (e.g., Karnataka, Maharashtra, Tamil Nadu).

*App opens and registered users grow steadily over time, indicating strong platform adoption.

*Some states show high app opens but lower transaction amounts, suggesting browsing or low-value usage.

📌 Insight: Adoption ≠ monetization. Engagement quality varies by region.

2️⃣ Transaction Type Distribution

Peer-to-peer payments and merchant payments dominate across most states.

Smaller states show limited transaction diversity, relying on 1–2 transaction types.

Seasonal and quarterly shifts are visible in transaction preferences.

📌 Insight: Users mature over time — they start with P2P and move toward merchant & bill payments.

3️⃣ Average Transaction Amount per User

States with fewer users sometimes have higher average transaction value per user.

High user count states often show lower per-user value, driven by micro-transactions.

📌 Insight: High-value users are not always in high-volume states.

4️⃣ Device Brand Analysis

A small number of mobile brands dominate registered users across all states.

Brand concentration varies regionally, indicating price sensitivity and device affordability impact adoption.

📌 Insight: Device ecosystem plays a key role in digital payment penetration.

5️⃣ Population & Density Impact

Population density has moderate correlation with transaction volume.

Highly dense districts show higher adoption, but not always higher transaction value.

📌 Insight: Urban density boosts adoption, but income & digital literacy drive transaction value.

6️⃣ Data Quality Validation

State-level metrics were validated by aggregating district-level data.

Minor differences were observed due to floating-point precision, not data errors.

📌 Insight: Data consistency is high — the dataset is reliable for decision-making.

🚀 Actionable Business Insights
📌 1. Improve Monetization in High-Adoption States

Target states with high app opens but low transaction value

Promote merchant payments, subscriptions, and bill payments

➡️ Action: Run state-specific campaigns to increase transaction value per user

📌 2. Upsell High-Value User Regions

Identify states with high average transaction amount per user

Introduce premium features, cashback tiers, or credit products

➡️ Action: Focus revenue initiatives on high-value but low-volume regions

📌 3. Device-Based Marketing Strategy

Optimize app performance for dominant device brands

Partner with OEMs for pre-installed apps or offers

➡️ Action: Align marketing with device brand dominance per state

📌 4. District-Level Expansion Strategy

Some districts show high population but low transaction penetration

Indicates untapped digital payment potential

➡️ Action: Invest in regional language campaigns & merchant onboarding

📌 5. Transaction Type Diversification

States dependent on limited transaction types risk stagnation

➡️ Action: Educate users about bill payments, investments, and insurance via the app
